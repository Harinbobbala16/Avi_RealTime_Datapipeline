```python
from pyspark.sql.functions import upper, col

print("Starting Silver Transformation...")
```

```python
df_bronze = spark.table("customers_bronze")
```

```python
df_silver = df_bronze.filter(col("city").isNotNull()) \
                     .withColumn("city", upper(col("city"))) \
                     .dropDuplicates(["customer_id"])

df_silver.write.format("delta").mode("overwrite").saveAsTable("customers_silver")

print("Silver table created!")
display(df_silver)
```
